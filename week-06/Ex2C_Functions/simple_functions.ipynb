```python
def favorite_things(name, movie, music):

    return f"""
My name is {name}
My favorite movie is {movie}
My favorite musician/band is {music}
""" 
```

```python
print(favorite_things( "Smaya",
                       "13 going on 30",
                       "Rauw Alejandro")) 
```

```python
def why_am_i_here(purpose):
    return purpose 
```

```python
print(why_am_i_here(
    "I joined the Year Up United Data Analyst program to build technical skills, gain career experience, and prepare for a successful future in data analytics."
))
```

```python
def favorite_place(place, why):
    return place, why
```

```python
 print(favorite_place("My favorite place is my room ","because I feel like time slows down and I feel more calm."))
```

```python
print(favorite_place(
    "Puerto Rico",
    "the beaches and culture are beautiful"
)) 
```

#The functions were created using parameters to make them reusable and easy to read. Using return statements made the functions more flexible and organized.
