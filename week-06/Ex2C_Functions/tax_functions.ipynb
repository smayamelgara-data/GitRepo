This lab demonstrates how to create and use Python functions to calculate different types of payroll taxes. 

The notebook includes:
- A function to calculate Social Security tax
- A function to calculate Medicare tax
- A function to calculate Federal withholding tax based on withholding codes

The lab also demonstrates:
- Using function parameters and return values
- Using if/else statements
- Testing functions with different employee pay examples


```python
def get_soc_sec_tax(gross_pay):
    tax = gross_pay * 0.062
    return tax
    
```

```python
def get_medicare_tax(gross_pay):
    tax = gross_pay * 0.0145
    return tax
    
```

```python
def get_federal_tax(gross_pay, withholding_code):

    if withholding_code == 0:
        tax_rate = 0.23

    elif withholding_code == 1:
        tax_rate = 0.21

    elif withholding_code == 2:
        tax_rate = 0.195

    elif withholding_code == 3:
        tax_rate = 0.185

    else:
        extra_withholding = withholding_code - 4
        tax_rate = 0.18 - (extra_withholding * 0.005)

    federal_tax = gross_pay * tax_rate

    return federal_tax
```

```python
gross_pay = 750
withholding_code = 0

print("Person 1")

print("Social Security Tax:",
      format(get_soc_sec_tax(gross_pay), ".2f"))

print("Medicare Tax:",
      format(get_medicare_tax(gross_pay), ".2f"))

print("Federal Tax:",
      format(get_federal_tax(gross_pay, withholding_code), ".2f"))
```
