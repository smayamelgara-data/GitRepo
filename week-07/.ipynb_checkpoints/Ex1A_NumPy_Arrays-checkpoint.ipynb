#Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

##Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])

print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through each row in the array, so it multiplies each row by 2 and prints the rows.

The second loop goes inside each row and prints each individual number multiplied by 2.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because even though y loops through each number,
# the code is multiplying x by 2, not y.
# Since x is the whole row, it prints the doubled row multiple times.
```

##Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# Error explanation:
# This gives an error because joined_array_1 and joined_array_2 have different shapes.
# NumPy cannot stack arrays unless they have the same shape.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# Error explanation:
# This gives an error because the arrays do not have the same number of rows.
# joined_array_1 has 4 rows, but arrA has 2 rows.
```

```python
h_stack = np.hstack((arrA, arrA))
# hstack stacks arrays horizontally, side by side.

v_stack = np.vstack((arrA, arrA))
# vstack stacks arrays vertically, one on top of the other.

d_stack = np.dstack((arrA, arrA))
# dstack stacks arrays depth-wise, creating a new third dimension.

print(h_stack)
print(v_stack)
print(d_stack)
```

##Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)

# This creates 2 arrays.
# Each array is 2-dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# This creates 3 arrays.
# Each array is 2-dimensional, but they are not all the same size.
```

##Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")

print(result_1)
```

##Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
