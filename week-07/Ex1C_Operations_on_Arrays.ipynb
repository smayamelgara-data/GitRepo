##Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)

r_2 = np.random.randint(1, 21, size=(1, 2))

r_3 = np.random.rand(2, 1)

r_4 = np.random.randint(20, 101, size=(4, 5))

r_5 = np.random.rand(4, 5)

print(r_1)
print(r_2)
print(r_3)
print(r_4)
print(r_5)
```

##Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)

print(r_4)
```

The shuffle function rearranged the rows of the array.  
The values inside each row stayed together, but the row order changed.

```python
new_shuffle = np.random.permutation(r_4)

print(new_shuffle)

print(r_4)
```

shuffle changes the original array directly.

permutation creates a shuffled copy while leaving the original array unchanged.

## Mathematical Operations

```python
print(r_1)
print(r_1.shape)

print(r_2)
print(r_2.shape)

result_1 = np.add(r_1, r_2)

print(result_1)
print(result_1.shape)
```

```python
print(r_1)
print(r_1.shape)

print(r_3)
print(r_3.shape)

result_2 = np.add(r_1, r_3)

print(result_2)
print(result_2.shape)
```

```python
print(r_2)
print(r_2.shape)

print(r_4)
print(r_4.shape)

result_3 = np.add(r_2, r_4)

print(result_3)
print(result_3.shape)
```

```python
print(np.add(r_1, r_4))

# This gives an error because the shapes are not compatible for broadcasting.
# r_1 has only 2 values, while r_4 has 5 columns.

print(np.add(r_2, r_4))
print(np.add(r_3, r_4))
```

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1,1,1,1,1]), r_4))
```

```python
print(np.add(np.array([[1],[1]]), r_4))

# Error explanation:
# This produces an error because the array shape does not match
# the number of rows in r_4.
```

Broadcasting allows NumPy to combine arrays with different shapes by stretching smaller arrays to fit larger ones.

When using add(), NumPy adds matching positions together.

The final shape depends on whether the arrays are compatible based on their dimensions and sizes.

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])

bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print("The first bag contains", bag1_total, "items, and the second bag contains", bag2_total, "items.")
```

```python
category_totals = np.add(bag1, bag2)

print(
    "There are",
    category_totals[0], "vegetables,",
    category_totals[1], "fruits,",
    category_totals[2], "boxes of snacks, and",
    category_totals[3], "beverages in the two grocery bags."
)
```

## Teamwork

```python
alexus = np.array([5, 4, 6, 4, 2])

print(alexus)
```

```python
blake = np.array([0, 5, 5, 3, 4])

print(blake)
```

```python
smaya = np.array([3, 6, 0, 3, 4])

print(smaya)
```

```python
all_knocks = np.array([alexus, blake, smaya])

print(all_knocks)
```

```python
person_totals = np.sum(all_knocks, axis=1)

group_total = np.sum(all_knocks)

print("Alexus knocked on", person_totals[0], "doors.")
print("Blake knocked on", person_totals[1], "doors.")
print("Smaya knocked on", person_totals[2], "doors.")
print("The group knocked on", group_total, "doors total.")
```

```python
daily_totals = np.sum(all_knocks, axis=0)

rolling_total = np.cumsum(daily_totals)

estimated_votes = rolling_total * 0.5

print(estimated_votes)
```
