# Northwind Review

## Categories
- Primary Key: CategoryID
- Represents: Product categories (Beverages, Seafood, Produce, etc.)
- Valuable in Power BI: Yes
- Why: Useful for grouping products and analyzing sales by category.
- Common Calculations: Sales by category, product counts by category.

### Key Columns
- CategoryID (Primary Key)
- CategoryName
- Description

---

## Territories
- Primary Key: TerritoryID
- Foreign Key: RegionID
- Represents: Geographic sales territories.
- Valuable in Power BI: Yes
- Why: Allows sales analysis by territory and region.
- Common Calculations: Sales by territory, sales by region.

### Key Columns
- TerritoryID
- TerritoryDescription
- RegionID

---

## Customers
- Primary Key: CustomerID
- Represents: Companies that place orders.
- Valuable in Power BI: Yes
- Why: Used to analyze customer purchasing behavior.
- Common Calculations:
  - Total orders per customer
  - Sales by country
  - Customer counts

### Key Columns
- CustomerID
- CompanyName
- ContactName
- City
- Country

---

## Employees
- Primary Key: EmployeeID
- Represents: Employees responsible for customer orders.
- Valuable in Power BI: Yes
- Why: Allows performance analysis of employees.
- Common Calculations:
  - Orders handled by employee
  - Sales by employee

### Key Columns
- EmployeeID
- FirstName
- LastName
- Title

---

## Employee Territories
- Primary Key: EmployeeID + TerritoryID
- Foreign Keys:
  - EmployeeID
  - TerritoryID
- Represents: Assignment of employees to territories.
- Valuable in Power BI: Yes
- Why: Connects employees to geographic sales areas.
- Common Calculations:
  - Employees per territory
  - Sales by territory and employee

### Key Columns
- EmployeeID
- TerritoryID

---

## Order Details
- Primary Key:
  - OrderID
  - ProductID
- Represents: Individual products included in each order.
- Valuable in Power BI: Very Important
- Why: Contains quantity, price, and discount information.
- Common Calculations:
  - Revenue
  - Quantity sold
  - Average discount

### Key Columns
- OrderID
- ProductID
- UnitPrice
- Quantity
- Discount

---

## Orders
- Primary Key: OrderID
- Foreign Keys:
  - CustomerID
  - EmployeeID
  - ShipVia
- Represents: Customer orders.
- Valuable in Power BI: Very Important
- Why: Connects customers, employees, and shipping information.
- Common Calculations:
  - Number of orders
  - Orders by month
  - Orders by customer

### Key Columns
- OrderID
- CustomerID
- EmployeeID
- OrderDate
- ShippedDate

---

## Overall Notes

- The Northwind database is highly normalized, meaning data is separated into multiple related tables to reduce duplication.
- Orders and Order Details are the most important tables for sales analysis.
- Customers, Products, Employees, Categories, Suppliers, and Shippers act as supporting lookup tables.
- Relationships between tables allow Power BI to create reports showing sales by customer, employee, product, category, supplier, territory, and region.
- The database structure is well suited for creating a star-schema style data model in Power BI.
